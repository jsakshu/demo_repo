
# 🔑 DBMS Keys with SQL Queries and Explanation

## 📘 Scenario: College Database

We will create two tables:
- `Department`
- `Student`

These will demonstrate:
- Primary Key
- Foreign Key
- Candidate Key
- Alternate Key
- Unique Key
- Super Key
- Composite Key (explained conceptually)

---

## 🧱 1. Create Table: Department

```sql
CREATE TABLE Department (
    dept_id INT PRIMARY KEY,                -- Primary Key (Candidate Key)
    dept_name VARCHAR(50) UNIQUE            -- Unique Key (Candidate Key)
);
```

- `dept_id`: **Primary Key** – uniquely identifies each department.
- `dept_name`: **Unique Key** – ensures department names are unique.

---

## 🧱 2. Create Table: Student

```sql
CREATE TABLE Student (
    student_id INT AUTO_INCREMENT PRIMARY KEY,   -- Surrogate + Primary Key
    student_roll VARCHAR(20) UNIQUE,             -- Unique Key (Candidate Key)
    name VARCHAR(50),
    email VARCHAR(100) UNIQUE,                   -- Alternate Key (Candidate Key)
    dept_id INT,                                 -- Foreign Key
    FOREIGN KEY (dept_id) REFERENCES Department(dept_id)
);
```

- `student_id`: Surrogate + Primary Key (auto-generated ID)
- `student_roll`: Unique Key, also a Candidate Key
- `email`: Alternate Key (candidate key not selected)
- `dept_id`: Foreign Key linking to Department table

## 📑 4. Explanation of Keys

| 🔑 Key Type      | 🧾 Table    | 🔣 Column(s)             | 📘 Explanation                                                                 |
|------------------|-------------|---------------------------|--------------------------------------------------------------------------------|
| Primary Key       | Department | `dept_id`                 | Uniquely identifies each department                                            |
| Primary Key       | Student    | `student_id`              | Auto-incremented, unique identifier for each student                          |
| Foreign Key       | Student    | `dept_id`                 | Links to `Department(dept_id)`                                                |
| Candidate Key     | Student    | `student_roll`, `email`   | Both can uniquely identify students                                           |
| Alternate Key     | Student    | `email`                   | Candidate key not chosen as primary key                                       |
| Unique Key        | Student    | `student_roll`, `email`   | Ensures unique values, may allow NULL once                                    |
| Super Key         | Student    | `student_id`, `email`, etc| Any key (or combination) that uniquely identifies a row                       |
| Surrogate Key     | Student    | `student_id`              | System-generated key (auto-incremented)                                       |
| Composite Key     | ✖️         | Not used here             | Would combine multiple columns to form a unique key (e.g., `order_id + item_id`) |

## 🔍 5. View Data with JOIN Query

```sql
SELECT 
    s.student_id, s.student_roll, s.name, s.email,
    d.dept_name
FROM 
    Student s
JOIN 
    Department d ON s.dept_id = d.dept_id;
```

## 📌 Summary

- **Primary Key** ensures each row is unique
- **Foreign Key** creates relationship between tables
- **Candidate Keys** are all possible unique identifiers
- **Alternate Key** is a candidate key not selected as primary
- **Unique Key** enforces uniqueness
- **Super Key** includes primary and other identifying combinations
- **Surrogate Key** is auto-generated (like `student_id`)
- **Composite Key** (not shown here) uses multiple columns together
