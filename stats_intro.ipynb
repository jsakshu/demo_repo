# 📘 Statistics Essentials 

---

## 📊 Slide 3: What is Statistics?

**Statistics** is a way of:
- Collecting information (like students' marks)
- Showing the information clearly (like charts or graphs)
- Making decisions or guesses based on it

**Two types of statistics:**
- **Descriptive Statistics:** Tells what the data says directly (like average marks)
- **Inferential Statistics:** Makes guesses for a large group using a small sample

**Example:** Ask 5 students their favorite ice cream. If most say “chocolate,” we guess that many others also like chocolate.

---

## 📋 Slide 4: Descriptive Statistics

Descriptive statistics help us explain the data using:
1. **Collecting data** – e.g., “What’s your favorite fruit?”
2. **Presenting data** – using a chart or table
3. **Summarizing data** – like finding the average

**Key Terms:**
- **Parameter:** Summary of a full group (population)
- **Statistic:** Summary of a small group (sample)

**Example:**
If you check the heights of 5 students and get an average of 150 cm, that’s a **statistic**.  
If you check every student in school, the average becomes a **parameter**.

---

## 🔍 Slide 5: Inferential Statistics

Helps us make guesses based on a sample.

**Two main uses:**
- **Estimation:** Guess the average height of all students by checking only 10
- **Hypothesis Testing:** Check if something we believe (like average marks being 80) is true

**Example:** Taste one spoon of soup to know if the whole bowl is tasty — just like checking a sample.

---

## 👨‍👩‍👧‍👦 Slide 6: Basic Terms – Population & Sample

- **Population:** Entire group (e.g., all students in the school)
- **Sample:** A part of the group (e.g., students in your class)
- **Variable:** What you’re measuring (e.g., height)
- **Data:** The actual values or answers (e.g., 150 cm, 155 cm...)

---

## 💡 Slide 7: More Key Terms

- **Random Variable:** Can store any value (e.g., numbers, names)
- **Experiment:** Planned activity to collect data (e.g., testing battery life)
- **Parameter:** A number that describes the whole population
- **Statistic:** A number that describes a sample

**Example:**
- Average height of all 500 students = **Parameter**
- Average height of 50 students = **Statistic**

---

## 📌 Slide 8: Examples

**Examples you can use:**

# 📂 Types of Data in Statistics

Data is generally classified into two main types:

---

## 1️⃣ Qualitative Data (Categorical Data)

This data describes **qualities or categories** — not measured with numbers.

### 📌 Types of Qualitative Data:

| Type         | Description                            | Example                        |
|--------------|----------------------------------------|--------------------------------|
| **Nominal**  | Categories with **no natural order**   | Gender: Male, Female, Other    |
| **Ordinal**  | Categories **with a meaningful order** | Shirt Size: Small, Medium, Large |

---

## 2️⃣ Quantitative Data (Numerical Data)

This data deals with **numbers and quantities** — things you can count or measure.

### 📌 Types of Quantitative Data:

| Type            | Description                               | Example                      |
|-----------------|-------------------------------------------|------------------------------|
| **Discrete**    | **Countable** numbers (whole numbers)     | Number of siblings: 0, 1, 2  |
| **Continuous**  | **Measurable** numbers (decimals possible)| Height: 165.5 cm, 172.3 cm   |

---

## 🧠 Simple Example

Imagine you're collecting data from students:

| Student | Gender | Shirt Size | Age | Number of Books | Height (cm) |
|---------|--------|------------|-----|------------------|--------------|
| A       | Male   | Medium     | 16  | 4                | 165.2        |
| B       | Female | Small      | 17  | 6                | 170.5        |

### Breakdown:
- **Gender** → Qualitative → **Nominal**
- **Shirt Size** → Qualitative → **Ordinal**
- **Age** → Quantitative → **Discrete**
- **Number of Books** → Quantitative → **Discrete**
- **Height** → Quantitative → **Continuous**

---

## 🎯 Summary Table

| Main Type     | Subtype     | Description                        | Example                |
|---------------|-------------|------------------------------------|------------------------|
| Qualitative   | Nominal     | No order                           | Colors, Gender         |
| Qualitative   | Ordinal     | Ordered categories                 | Rank, Shirt size       |
| Quantitative  | Discrete    | Countable, whole numbers           | No. of students        |
| Quantitative  | Continuous  | Measurable, can have decimals      | Weight, Temperature    |


# 📘 Data Collection Examples in Statistics

---

## 📱 Example 1: Mobile Phone Usage Among Students

### ✅ Step 1: Define the Aim
- Estimate the **average number of hours** students use mobile phones per day.

### ✅ Step 2: Define Variable & Population
- **Variable:** Daily mobile phone usage time (in hours)  
- **Population:** All college students in the city

### ✅ Step 3: Data Collection & Measurement
- **Sampling Procedure:** Random sampling from different colleges  
- **Sample Size:** 150 students  
- **Measurement Method:** Use phone’s screen-time report or self-reporting via survey

### ✅ Step 4: Analysis Techniques
- **Descriptive:** Mean, minimum, maximum, standard deviation of phone usage  
- **Inferential:** Confidence interval to predict usage for all students, or t-test by gender or course

---

## 💡 Example 2: Life of an Electronic Bulb

### ✅ Step 1: Define the Aim
- Estimate the **average lifespan** of an electronic component (e.g., LED bulb).

### ✅ Step 2: Define Variable & Population
- **Variable:** Lifespan (in hours), power rating, usage  
- **Population:** All LED bulbs manufactured in the factory

### ✅ Step 3: Data Collection & Measurement
- **Sampling Procedure:** Random sampling from different batches  
- **Sample Size:** 300 bulbs  
- **Measurement Method:** Use a digital timer to track when each bulb fails

### ✅ Step 4: Analysis Techniques
- **Descriptive:** Average lifespan, range, and variability  
- **Inferential:** Use confidence interval or hypothesis test to estimate population lifespan

---

## 🧍 Example 3: BMI of College Students

### ✅ Step 1: Define the Aim
- Estimate the **average Body Mass Index (BMI)** of college students.

### ✅ Step 2: Define Variable & Population
- **Variable:** BMI = Weight (kg) / [Height (m)]²  
- **Population:** All college students in a university or city

### ✅ Step 3: Data Collection & Measurement
- **Sampling Procedure:** Stratified sampling from various departments  
- **Sample Size:** 200 students  
- **Measurement Method:** Weighing scale and measuring tape

### ✅ Step 4: Analysis Techniques
- **Descriptive:** Calculate mean BMI, SD, max, and min  
- **Inferential:** Estimate average BMI using confidence interval; compare groups using t-tests



# 📊 Sampling Methods Explained with One Common Example

## 📱 Scenario:
**Objective:** To study the *average daily screen time* of college students.

---

## 🔷 1. Probability Sampling Methods

### 🟢 a) Simple Random Sampling
- **Explanation:** Every student has an equal chance of being selected.
- **Example:** From a list of 500 students, randomly pick 50 using a lottery method or random number generator.

---

### 🟢 b) Systematic Sampling
- **Explanation:** Select every *k-th* student from a list.
- **Example:** Arrange all 500 students alphabetically. Pick every 10th student (10th, 20th, 30th...).

---

### 🟢 c) Stratified Sampling
- **Explanation:** Divide the population into subgroups (strata), then randomly sample from each.
- **Example:** Divide students into 3 courses — Science, Commerce, and Arts. Randomly pick 20 from each course.

---

### 🟢 d) Cluster Sampling
- **Explanation:** Divide population into clusters, then randomly choose entire clusters.
- **Example:** Out of 10 colleges in your city, randomly select 3 colleges. Survey **all students** in those colleges.

---

## 🔶 2. Non-Probability Sampling Methods

### 🔴 a) Convenience Sampling
- **Explanation:** Select people who are easiest to reach.
- **Example:** Survey 50 students who are available in the canteen or library.

---

### 🔴 b) Judgmental (Purposive) Sampling
- **Explanation:** Select based on specific criteria or judgment.
- **Example:** Select only students who use phones more than 5 hours a day.

---

### 🔴 c) Quota Sampling
- **Explanation:** Choose a fixed number from each category, not randomly.
- **Example:** Decide to survey 25 boys and 25 girls, but pick whoever is available.

---

### 🔴 d) Snowball Sampling
- **Explanation:** Existing participants refer new participants.
- **Example:** Start with 5 students. Each refers 2 more friends who use phones heavily.

---

| Sampling Type           | Random? | How It Works                                                  | Example                                               |
