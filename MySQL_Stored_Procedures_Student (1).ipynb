# 📘 Introduction to Stored Procedures in MySQL

A **stored procedure** is a set of SQL statements with an assigned name that's stored in the database. It can be reused, executed repeatedly, and is useful for encapsulating logic and improving performance.

## ✅ Benefits of Stored Procedures
- **Reusability**: Write once, use many times.
- **Performance**: Precompiled and faster for repeated execution.
- **Security**: Grant users permission to execute procedures without exposing the underlying tables.
- **Modularity**: Keeps SQL logic centralized and easier to manage.
- **Maintainability**: Changes only need to be made in one place.

## 📦 Parameter Types in MySQL Stored Procedures
| Type     | Description                              |
|----------|------------------------------------------|
| `IN`     | Default. Passes a value **into** a procedure. |
| `OUT`    | Passes a value **out** of a procedure.       |
| `INOUT`  | Passes a value **in**, and the procedure can **modify and return** it. |

## 🔧 General Syntax
```sql
DELIMITER //
CREATE PROCEDURE procedure_name (
    [IN | OUT | INOUT] param_name datatype,
    ...
)
BEGIN
    -- SQL statements
END //
DELIMITER ;
```

This notebook uses a sample `Student` table to demonstrate all three parameter types and the typical use cases of stored procedures in MySQL.

# 📘 Stored Procedures in MySQL with `Student` Table
This notebook demonstrates how to create and use stored procedures in **MySQL** using a `Student` table with `IN`, `OUT`, and `INOUT` parameters.

## 🧾 Student Table Schema
```sql
CREATE TABLE Student (
    student_roll VARCHAR(10) PRIMARY KEY,
    name VARCHAR(100),
    email VARCHAR(100),
    dept_id INT
);
```

## 📥 IN Parameter
**Procedure**: Get student details by roll number
```sql
DELIMITER //
CREATE PROCEDURE GetStudentByRoll (
    IN p_roll VARCHAR(10)
)
BEGIN
    SELECT * FROM Student WHERE student_roll = p_roll;
END //
DELIMITER ;
```
**Usage:**
```sql
CALL GetStudentByRoll('CS01');
```

## 📤 OUT Parameter
**Procedure**: Get student email by roll number
```sql
DELIMITER //
CREATE PROCEDURE GetStudentEmail (
    IN p_roll VARCHAR(10),
    OUT p_email VARCHAR(100)
)
BEGIN
    SELECT email INTO p_email FROM Student WHERE student_roll = p_roll;
END //
DELIMITER ;
```
**Usage:**
```sql
CALL GetStudentEmail('CS01', @student_email);
SELECT @student_email;
```

## 🔁 INOUT Parameter
**Procedure**: Update department ID of a student
```sql
DELIMITER //
CREATE PROCEDURE UpdateStudentDept (
    INOUT p_roll VARCHAR(10),
    IN p_new_dept INT
)
BEGIN
    UPDATE Student SET dept_id = p_new_dept WHERE student_roll = p_roll;
END //
DELIMITER ;
```
**Usage:**
```sql
SET @roll = 'CS01';
CALL UpdateStudentDept(@roll, 5);
```

## ➕ Add Student with IN Parameters
**Procedure**: Insert new student
```sql
DELIMITER //
CREATE PROCEDURE AddStudent (
    IN p_roll VARCHAR(10),
    IN p_name VARCHAR(100),
    IN p_email VARCHAR(100),
    IN p_dept_id INT
)
BEGIN
    INSERT INTO Student (student_roll, name, email, dept_id)
    VALUES (p_roll, p_name, p_email, p_dept_id);
END //
DELIMITER ;
```
**Usage:**
```sql
CALL AddStudent('CS10', 'Henry', 'henry@example.com', 1);
```

## 🧹 Drop Existing Procedure
**Tip:** Use this if you need to re-create a procedure that already exists.
```sql
DROP PROCEDURE IF EXISTS AddStudent;
```
